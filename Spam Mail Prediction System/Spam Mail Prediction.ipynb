# Importing the dependencies

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Data collection and Analysis

```python
dataset = pd.read_csv("/content/spam_ham_dataset.csv")
dataset.head(3)
```

```python
# seperating the data as texts and label
x = dataset['text']
y = dataset['label_num']
```

```python
x.head(5)
```

```python
y.head(5)
```

# Spliting the dataset into training and testing

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

Feature Extraction

```python
# transform the text data to feature vector so that can be used as input to the logistic regression model
from sklearn.feature_extraction.text import TfidfVectorizer
feature_extraction = TfidfVectorizer(min_df=1, stop_words = "english", lowercase=True)

x_train_features = feature_extraction.fit_transform(x_train)
x_test_features = feature_extraction.transform(x_test)

# convsert y_train and y_test values as integers
y_train = y_train.astype('int')
y_test = y_test.astype('int')
#
```

```python
print(x_train_features)
```

# Model Training

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(x_train_features, y_train)
```

# Evaluating the model

```python
model.score(x_train_features, y_train)
```

```python
model.score(x_test_features, y_test)
```

# Build a Predictive System

```python
input = ["re : tenaska iv 10 / 00 we have received all of the money from the spot sales for tenaska iv in october 2000 , except for the tenaska iv sale and the $ 0 . 04 fee . megan"]

# convert text to features vector
input_features = feature_extraction.transform(input)

# making prediction
prediction = model.predict(input_features)
print(prediction)

if prediction[0] == 0:
  print("Ham mail")
else:
  print("Spam mail")
```
