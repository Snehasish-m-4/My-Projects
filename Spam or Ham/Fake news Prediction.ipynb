```python
import re                           # use to help to read text data from the document
from nltk.corpus import stopwords       # important content of the text
from nltk.stem.porter import PorterStemmer   # it removes the prefix and sufix from the root word
from sklearn.feature_extraction.text import TfidfVectorizer
```

# Data PreProcessing

```python
df.isnull().sum()
```

```python
# replaceing missing values with Null string
df = df.fillna('')
```

```python
# Combining the title and text into one single column
df['content'] = df['title']+' '+df['text']
```

```python
df.head(5)
```

```python
# Spliting data into input and output data
x = df.drop(columns='label', axis=1)
y = df['label']
```

Steaming:

steaming is not but the process of reduceing the word to its root word.

example:
actor, acting,etc ---> act

```python
porter_steam = PorterStemmer()
```

```python
def steaming (content):
  steamed_content = re.sub('[^a-zA-Z]',' ',content)
  steamed_content = steamed_content.lower()
  steamed_content = steamed_content.split()
  steamed_content = [porter_steam.stem(word) for word in steamed_content if not word in stopwords.words('english')]
  steamed_content = ' '.join(steamed_content)
  return steamed_content
```

```python
# Apply steaming to all the content columns
df['content'] = df['content'].apply(steaming)
```

```python
print(df['content'])
```

```python
# seperating data and levels
x = df['content'].values
y = df['label'].values
```

```python
# converting the textual data into numerical data using vectorization
vectorizer = TfidfVectorizer()
vectorizer.fit(x)

x = vectorizer.transform(x)
```

```python
print(x)
```

```python
print(y)
```

# Spliting the data into train and test data

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

# Modeling training

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(x_train, y_train)
```

Model Evaluation

```python
lr.score(x_test, y_test)
```

```python
lr.score(x_train, y_train)
```

# Making a Predictive system

```python
X_new = x_test[111]

prediction = lr.predict(X_new)
print(prediction)

if (prediction[0]==0):
  print('The news is Fake')
else:
  print('The news is Real')
```
