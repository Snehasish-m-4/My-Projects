0 --> Normal transaction

1 --> Fraud transaction

```python
# seperating the data for analysis
legit = dataset[dataset.Class == 0]
fraud = dataset[dataset.Class == 1]
```

```python
legit.shape, fraud.shape
```

# Under Sampling

Build a sample dataset containing similar distribution of normal and fraud data point

we take 492 random data from the legit dataset and combine them with the fraud dataset

```python
# select random 492 datapoints
legit_sample = legit.sample(n=211)
```

### Concatenting the datasets

```python
df = pd.concat([legit_sample, fraud], axis=0)
```

```python
df.head(3)
```

```python
df.shape
```

```python
df['Class'].value_counts()
```

# Spliting the dataset into input and output data

```python
x = df.iloc[:,:-1]
y = df['Class']
```

# Spliting into training and testing data

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state=42)
```

# Model training

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
```

```python
lr.fit(x_train, y_train)
```

# Model Evaluation

```python
lr.score(x_train, y_train)
```

```python
lr.score(x_test, y_test)
```

### Making a predictive system

```python
input = (x_test.iloc[[63]])

# Convert the input data into a numpy array
input_array = np.asarray(input)

# Reshaping the numpy array as we predict for only one instance
input_reshaped = input_array.reshape(1,-1)

prediction = lr.predict(input_reshaped)
print(prediction)

if(prediction[0] == 0):
    print("Normal Transaction")
else:
    print("Fraud Transaction")
```
