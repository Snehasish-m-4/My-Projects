# Importing the Dependencies

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import difflib
```

# Data Collection and Analysis

```python
dataset = pd.read_csv("/content/movies.csv")
dataset.head(3)
```

```python
dataset.info()
```

```python
dataset.isnull().sum()
```

```python
dataset.shape
```

```python
# select the relevent features from the columns
slected_features = ['genres','keywords','tagline','title','cast','director']
print(slected_features)
```

```python
# replaceing the empty places with the null string
for feature in slected_features:
  dataset[feature] = dataset[feature].fillna('')
```

```python
dataset.isnull().sum()
```

```python
# combine all 6 features
combined_features = dataset['genres']+' '+dataset['keywords']+' '+dataset['tagline']+' '+dataset['title']+" "+dataset['cast']+" "+dataset['director']
```

```python
# convert the text data to features vector
from sklearn.feature_extraction.text import TfidfVectorizer
vec = TfidfVectorizer()

feature_vector = vec.fit_transform(combined_features)
```

```python
print(feature_vector)
```

### Cosine Similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(feature_vector)
```

```python
print(similarity)
```

```python
print(similarity.shape)
```

## Recommended similar movies

```python
# getting inout from the user
movie_name = input("Enter the movie name : ")
```

```python
# creating a list with all the movies name on the list
list_of_all_titles = dataset['title'].tolist()
print(list_of_all_titles)
```

```python
# find the close match for the movie given by the user
find_close_match = difflib.get_close_matches(movie_name,list_of_all_titles)
print(find_close_match)
```

```python
close_match = find_close_match[0]
print(close_match)
```

```python
# find the index of the movie from the title
index_of_the_movie = dataset[dataset.title == close_match]['index'].values[0]
print(index_of_the_movie)
```

```python
# getting the list of similar movies
similarity_score = list(enumerate(similarity[index_of_the_movie]))
print(similarity_score)
```

```python
len(similarity_score)
```

```python
# sorting the movies based on their similarity score
sorted_similar_movies = sorted(similarity_score,key = lambda x:x[1],reverse = True)
print(sorted_similar_movies)
```

```python
# print the name of the movies based on the indexes
print("Movies suggested for you : \n")

i = 1
for movie in sorted_similar_movies:
  index = movie[0]
  title_from_index = dataset[dataset.index == index]['title'].values[0]
  if(i<15):
    print(i,'.',title_from_index)
    i+=1
```

# Movie Recommendation System

```python
# getting inout from the user
movie_name = input("Enter the movie name : ")

# creating a list with all the movies name on the list
list_of_all_titles = dataset['title'].tolist()

# find the close match for the movie given by the user
find_close_match = difflib.get_close_matches(movie_name,list_of_all_titles)

# find the index of the movie from the title
close_match = find_close_match[0]
index_of_the_movie = dataset[dataset.title == close_match]['index'].values[0]

# getting the list of similar movies
similarity_score = list(enumerate(similarity[index_of_the_movie]))

# sorting the movies based on their similarity score
sorted_similar_movies = sorted(similarity_score,key = lambda x:x[1],reverse = True)

# print the name of the movies based on the indexes
print("Movies suggested for you : \n")

i = 1
for movie in sorted_similar_movies:
  index = movie[0]
  title_from_index = dataset[dataset.index == index]['title'].values[0]
  if(i<15):
    print(i,'.',title_from_index)
    i+=1
```
